# NLP com spaCy

## Doc e Tokens

```python
# Importando spacy (pip install spacy)
import spacy

# Carregando o modelo (python -m spacy download pt_core_newa_lg)
nlp = spacy.load("pt_core_news_lg")
```

```python
# Processando um texto
texto = "Maria trabalha na Microsoft."

doc = nlp(texto)
print(type(doc))
print(doc)
```

```python
# Analisando o doc
dados = doc.to_dict()
for k, v in dados.items():
    print(f"Chave: {k:25} | Valor: {type(v)}")
```

```python
# Percorrendo palavra por palavra 
for token in doc:
    print(token)
```

```python
# Quantidade de Tokens
len(doc)
```

```python
# Acessando um Token
print(doc[0])

# Texto original
print(doc.text)
```

```python
# Comparando com NLTK
import nltk
from nltk.tokenize import word_tokenize

tokens = word_tokenize(texto)
print(tokens)

# spaCy
doc = nlp(texto)
for token in doc:
    print(token.text)
```

## Atributos dos Tokens

```python
token = doc[4]
print(token)

# Verifica se é composto apenas por letras (alfabético)
print(token.is_alpha)

# Verifica se é um stopword
print(token.is_stop)

# Verifica se é pontução
print(token.is_punct)
```

```python
# Formato da palavra

for token in doc:
    print(token, token.shape_)
```

## Separação de Frases (Sentence Segmentation)

```python
texto = """
Hoje fez sol!
Amanhã poderá chover.
Depois voltará a fazer calor."""

doc = nlp(texto)
```

```python
for frase in doc.sents:
    print(frase)
```

## Lematização 

```python
# Lematizar significa encontrar a forma "base' de uma palavra
texto = "As crianças estavam correndo felizes."
doc = nlp(texto)
for token in doc:
    print(token.text, "->", token.lemma_)
```

## Classes gramaticais

```python
texto = "O menino correu rapidamente para casa."
doc = nlp(texto)

for token in doc:
    print(token.text, "->", token.pos_)
```

## Entidades

```python
texto = "João trabalha na Microsoft em São Paulo desde janeiro de 2024."
doc = nlp(texto)

for entidade in doc.ents:
    print(entidade.text, "->", entidade.label_)
```

# Vetores de semelhança

```python
nlp = spacy.load("en_core_web_lg")

pairs = [
    ("buy", "purchase"),
    ("dog", "cat"),
    ("Brazil", "Argentina"),
    ("computer", "keyboard"),
    ("banana", "airplane")
]

for word1, word2 in pairs:
    doc1 = nlp(word1)
    doc2 = nlp(word2)

    print(f"{word1:10} x {word2:10} Semelhança: {doc1.similarity(doc2)*100:.2f}%")
```
