# Exemplo 2: Classificação (Previssão de Espécies de Flores)

# Dados

```python
from sklearn.datasets import load_iris
dados = load_iris()
```

```python
# Atributos
X = dados.data # Características (tamanho da pétala, sépala etc...)

# Alvo
y = dados.target
```

```python
# Características
dados.feature_names
```

```python
# Classes
dados.target_names
```

```python
# Tabela (DataFrame)
import pandas as pd

df = pd.DataFrame(dados.data, columns=dados.feature_names)
df['target'] = y
df['target_name'] = df['target'].apply(lambda X: dados.target_names[X])
df
```

```python
df['target_name'].value_counts()
```

# Divisão treino/teste

```python
from sklearn.model_selection import train_test_split
# 70% treino / 30% teste
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Modelo

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
# Modelo
modelo = DecisionTreeClassifier()

# Treinamento
modelo.fit(X_treino, y_treino)
```

# Previsão com os dados de teste

```python
previsoes = modelo.predict(X_teste)
previsoes
```

```python
# Tabela de comparação
df_comp = pd.DataFrame(X_teste, columns=dados.feature_names)
df_comp['target'] = y_teste
df_comp['previsao'] = previsoes
df_comp
```

# Avaliação

```python
from sklearn.metrics import accuracy_score
```

```python
acuracia = accuracy_score(y_teste, previsoes)
print(f'O modelo aceertou {acuracia * 100:.2f}% das previsões.')
```

# Plot da árvore

```python
import matplotlib.pyplot as plt
```

```python
plt.Figure(figsize=(18,10))
plot_tree(modelo, 
          feature_names=dados.feature_names,
          class_names=dados.target_names,
          filled=True,
          rounded=True,
          fontsize=8)
plt.show()
```
